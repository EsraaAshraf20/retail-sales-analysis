2. Data Cleaning & Validation

In this section, we validate data types, check missing values and duplicate records, and verify the consistency of transaction amounts.

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
# 1. Convert Date column to datetime
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

# 2. Check duplicate Transaction IDs
print("Duplicate Transaction IDs:")
print(df["Transaction ID"].duplicated().sum())

# 3. Check if Total Amount is calculated correctly
df["Expected Amount"] = df["Quantity"] * df["Price per Unit"]

print("\nIncorrect Total Amount Rows:")
print((df["Total Amount"] != df["Expected Amount"]).sum())

# 4. Display updated data types
print("\nUpdated Data Types:")
print(df.dtypes)
```

3. Exploratory Data Analysis (EDA)

This section explores retail sales data to identify sales patterns, compare product categories, understand customer demographics, and investigate purchasing behavior.

```python
# Remove the temporary column
df.drop(columns=["Expected Amount"], inplace=True)

# Basic statistics
print("Dataset Summary:")
display(df.describe())

# Total Sales
print("\nTotal Sales:")
print(df["Total Amount"].sum())

# Average Transaction Value
print("\nAverage Transaction Value:")
print(df["Total Amount"].mean())

# Sales by Product Category
category_sales = (
    df.groupby("Product Category")["Total Amount"]
    .sum()
    .sort_values(ascending=False)
)

print("\nSales by Product Category:")
display(category_sales)

# Number of transactions in each category
print("\nTransactions by Product Category:")
display(df["Product Category"].value_counts())
```

```python


# Prepare sales by category
category_sales = (
    df.groupby("Product Category", as_index=False)["Total Amount"]
    .sum()
    .sort_values("Total Amount", ascending=False)
)

# Create the chart
fig = px.bar(
    category_sales,
    x="Product Category",
    y="Total Amount",
    title="Total Sales by Product Category",
    labels={
        "Product Category": "Product Category",
        "Total Amount": "Total Sales"
    },
    text="Total Amount"
)

fig.update_traces(texttemplate="%{text:,.0f}", textposition="outside")

fig.update_layout(
    template="plotly_white",
    yaxis_title="Total Sales",
    xaxis_title="Product Category"
)

fig.show()
```

Insight: Sales Performance by Product Category

Electronics generated the highest total sales at 156,905, closely followed by Clothing at 155,580. Beauty recorded the lowest total sales at 143,515. This suggests that Electronics and Clothing are the leading categories in terms of revenue, although profitability cannot be determined without cost data.

```python


# Calculate monthly sales
monthly_sales = (
    df.groupby(pd.Grouper(key="Date", freq="MS"))["Total Amount"]
    .sum()
    .reset_index()
)

# Create a line chart
fig = px.line(
    monthly_sales,
    x="Date",
    y="Total Amount",
    title="Monthly Sales Trend",
    labels={
        "Date": "Month",
        "Total Amount": "Total Sales"
    },
    markers=True
)

fig.update_layout(
    template="plotly_white",
    xaxis_title="Month",
    yaxis_title="Total Sales"
)

fig.show()

# Find the month with the highest sales
best_month = monthly_sales.loc[monthly_sales["Total Amount"].idxmax()]

print("Highest Sales Month:", best_month["Date"].strftime("%B %Y"))
print("Sales:", best_month["Total Amount"])
```

Monthly Sales Trend Insight:

Monthly sales fluctuated throughout the analyzed period. May 2023 recorded the highest sales, reaching 53,150, while September 2023 recorded the lowest sales among the complete months, at 23,620. This indicates variations in monthly sales performance over time.

Note: January 2024 includes data for only one day, so it should not be compared directly with full months.

```python
import plotly.express as px

category_sales = (
    df.groupby("Product Category", as_index=False)["Total Amount"]
    .sum()
)

fig = px.pie(
    category_sales,
    names="Product Category",
    values="Total Amount",
    title="Sales Distribution by Product Category",
    hole=0.35
)

fig.update_traces(
    textinfo="label+percent",
    hovertemplate="%{label}<br>Sales: %{value:,.0f}<br>Share: %{percent}<extra></extra>"
)

fig.update_layout(template="plotly_white")
fig.show()
```

Electronics contributed the largest share of total sales, followed closely by Clothing. Beauty had the smallest share. Overall, sales revenue was distributed relatively evenly across the three product categories.

```python
fig = px.histogram(
    df,
    x="Age",
    nbins=10,
    title="Customer Age Distribution",
    labels={"Age": "Customer Age"},
    color_discrete_sequence=["#636EFA"]
)

fig.update_layout(
    template="plotly_white",
    xaxis_title="Age",
    yaxis_title="Number of Transactions"
)

fig.show()

print("Average Customer Age:", round(df["Age"].mean(), 2))
print("Median Customer Age:", df["Age"].median())
```

Customer Age Distribution Insight:

The average customer age is 41.39 years, while the median age is 42 years. Customers range in age from 18 to 64 years, indicating that the dataset includes a broad range of age groups, with the typical customer being around 41–42 years old.

This suggests that the retail business serves customers across different age groups rather than focusing on a single age range.

```python
fig = px.box(
    df,
    x="Product Category",
    y="Total Amount",
    color="Product Category",
    title="Transaction Value by Product Category",
    labels={
        "Product Category": "Product Category",
        "Total Amount": "Transaction Value"
    },
    points="outliers"
)

fig.update_layout(
    template="plotly_white",
    showlegend=False
)

fig.show()
```

Transaction Value by Product Category Insight:

The box plot shows considerable variation in transaction values across all three product categories. In each category, transaction values range from low amounts to high-value purchases, with several outliers reaching up to 2,000. This indicates that customers make purchases of different values across Beauty, Clothing, and Electronics.

Further analysis of the median and interquartile range can help identify differences in typical transaction values between categories.

```python
fig = px.scatter(
    df,
    x="Quantity",
    y="Total Amount",
    color="Product Category",
    hover_data=["Price per Unit", "Gender", "Age"],
    title="Quantity vs. Total Transaction Amount",
    labels={
        "Quantity": "Quantity Purchased",
        "Total Amount": "Total Amount"
    }
)

fig.update_layout(
    template="plotly_white",
    xaxis_title="Quantity Purchased",
    yaxis_title="Total Transaction Amount"
)

fig.show()

print(
    "Correlation between Quantity and Total Amount:",
    round(df["Quantity"].corr(df["Total Amount"]), 3)
)
```

Quantity vs. Transaction Amount: The scatter plot shows a moderate positive relationship between quantity purchased and total transaction amount (correlation = 0.374). However, the relationship is not strong because product prices also affect the total amount spent.

```python
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# ==========================================
# 1. Load the dataset
# ==========================================
df = pd.read_csv("retail_sales_dataset.csv")

# ==========================================
# 2. Clean the data
# ==========================================
df.columns = df.columns.str.strip()

required_columns = [
    "Transaction ID",
    "Date",
    "Customer ID",
    "Product Category",
    "Total Amount"
]

missing_columns = [
    column for column in required_columns
    if column not in df.columns
]

if missing_columns:
    raise ValueError(f"Missing columns: {missing_columns}")

df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
df["Total Amount"] = pd.to_numeric(
    df["Total Amount"], errors="coerce"
)

df = df.dropna(
    subset=[
        "Transaction ID",
        "Date",
        "Customer ID",
        "Product Category",
        "Total Amount"
    ]
).copy()

if df.empty:
    raise ValueError("The dataset contains no valid rows.")

# ==========================================
# 3. Calculate KPIs
# ==========================================
total_sales = df["Total Amount"].sum()
total_transactions = df["Transaction ID"].nunique()
avg_transaction = df["Total Amount"].mean()
unique_customers = df["Customer ID"].nunique()

# ==========================================
# 4. Sales by product category
# ==========================================
category_sales = (
    df.groupby("Product Category", as_index=False)["Total Amount"]
    .sum()
    .sort_values("Total Amount", ascending=False)
)

# ==========================================
# 5. Monthly sales
# ==========================================
monthly_sales = (
    df.groupby(pd.Grouper(key="Date", freq="MS"))["Total Amount"]
    .sum()
    .reset_index()
)

# Keep complete months in 2023
monthly_complete = monthly_sales[
    (monthly_sales["Date"] >= pd.Timestamp("2023-01-01"))
    & (monthly_sales["Date"] < pd.Timestamp("2024-01-01"))
].copy()

# ==========================================
# 6. Create dashboard layout
# ==========================================
fig = make_subplots(
    rows=3,
    cols=2,
    specs=[
        [{"type": "indicator"}, {"type": "indicator"}],
        [{"type": "indicator"}, {"type": "indicator"}],
        [{"type": "bar"}, {"type": "scatter"}]
    ],
    subplot_titles=(
        "Total Sales",
        "Total Transactions",
        "Average Transaction Value",
        "Unique Customers",
        "Sales by Product Category",
        "Monthly Sales Trend (2023)"
    ),
    vertical_spacing=0.10,
    horizontal_spacing=0.12
)

# ==========================================
# 7. Add KPI cards
# ==========================================
kpis = [
    (total_sales, "Total Sales", 1, 1, ",.0f"),
    (total_transactions, "Total Transactions", 1, 2, ",.0f"),
    (avg_transaction, "Average Transaction Value", 2, 1, ",.2f"),
    (unique_customers, "Unique Customers", 2, 2, ",.0f")
]

for value, label, row, col, number_format in kpis:
    fig.add_trace(
        go.Indicator(
            mode="number",
            value=value,
            title={"text": label},
            number={
                "valueformat": number_format,
                "font": {"size": 30, "color": "#243B53"}
            }
        ),
        row=row,
        col=col
    )

# ==========================================
# 8. Add category sales bar chart
# ==========================================
colors = ["#4263EB", "#20A39E", "#F4A261"]

fig.add_trace(
    go.Bar(
        x=category_sales["Product Category"],
        y=category_sales["Total Amount"],
        marker_color=colors[:len(category_sales)],
        text=category_sales["Total Amount"],
        texttemplate="%{text:,.0f}",
        textposition="outside",
        cliponaxis=False,
        hovertemplate=(
            "<b>%{x}</b><br>"
            "Total Sales: %{y:,.0f}"
            "<extra></extra>"
        )
    ),
    row=3,
    col=1
)

# ==========================================
# 9. Add monthly sales trend
# ==========================================
fig.add_trace(
    go.Scatter(
        x=monthly_complete["Date"],
        y=monthly_complete["Total Amount"],
        mode="lines+markers",
        line={"color": "#4263EB", "width": 3},
        marker={"size": 8, "color": "#20A39E"},
        hovertemplate=(
            "<b>%{x|%B %Y}</b><br>"
            "Total Sales: %{y:,.0f}"
            "<extra></extra>"
        )
    ),
    row=3,
    col=2
)

# ==========================================
# 10. Dashboard appearance
# ==========================================
fig.update_layout(
    title={
        "text": (
            "Retail Sales Dashboard"
            "<br><sup>Sales Performance Overview | 2023</sup>"
        ),
        "x": 0.5,
        "xanchor": "center",
        "font": {"size": 25, "color": "#243B53"}
    },
    template="plotly_white",
    height=850,
    showlegend=False,
    margin={"l": 55, "r": 35, "t": 140, "b": 80},
    paper_bgcolor="#FFFFFF",
    plot_bgcolor="#FFFFFF",
    font={"family": "Arial", "size": 12, "color": "#334155"}
)

# ==========================================
# 11. Format chart axes
# ==========================================
fig.update_yaxes(
    title_text="Total Sales",
    tickformat=",.0f",
    gridcolor="#E8EDF3",
    row=3,
    col=1
)

fig.update_yaxes(
    title_text="Total Sales",
    tickformat=",.0f",
    gridcolor="#E8EDF3",
    row=3,
    col=2
)

fig.update_xaxes(
    showgrid=False,
    row=3,
    col=1
)

fig.update_xaxes(
    tickformat="%b",
    showgrid=False,
    row=3,
    col=2
)

# ==========================================
# 12. Display dashboard
# ==========================================
fig.show()

print(f"Total Sales: {total_sales:,.0f}")
print(f"Total Transactions: {total_transactions:,}")
print(f"Average Transaction Value: {avg_transaction:,.2f}")
print(f"Unique Customers: {unique_customers:,}")
print(
    "Note: January 2024 is excluded because it contains "
    "data for only one day."
)
```

Key Findings, Business Recommendations & Limitations

1. Key Findings

Overall Sales Performance

* Total sales reached 456,000 across 1,000 transactions.
* The average transaction value was 456.

Product Category Performance

* Electronics generated the highest revenue at 156,905, narrowly exceeding Clothing (155,580).
* Beauty recorded the lowest revenue at 143,515, making it a potential area for further investigation.
* The relatively small gap between Electronics and Clothing suggests that both categories contribute substantially to total sales.

Monthly Sales Performance

* May 2023 was the strongest complete month, with sales of 53,150.
* September 2023 recorded the lowest sales among complete months, at 23,620.
* These fluctuations highlight opportunities to investigate seasonal patterns and changes in customer demand.

Customer Demographics

* Customer ages ranged from 18 to 64, with an average age of approximately 41 years.
* This indicates that the dataset includes customers from a broad adult age range.

Purchase Behavior

* Quantity purchased showed a moderate positive correlation with transaction value (r = 0.374).
* This suggests that larger quantities are generally associated with higher transaction values, although unit prices also play a role.

2. Business Recommendations

1. Investigate Monthly Performance: Examine possible reasons for the sales peak in May and the decline in September, using additional information such as promotions, product availability, and seasonal demand.
2. Evaluate the Beauty Category: Analyze product-level sales, prices, and quantities to identify opportunities to improve category performance.
3. Optimize Product and Pricing Decisions: Compare unit prices and quantities sold to understand which products contribute most to sales revenue.
4. Improve Customer Analysis: Collect customer purchase history over time to identify repeat customers, measure retention, and understand purchasing patterns.

3. Data Limitations

* Revenue vs. Profit: Product costs are unavailable, so profit margins and overall profitability cannot be calculated.
* Incomplete Final Month: January 2024 contains records for only one day and should not be compared directly with full months.
* Limited Customer History: Each transaction has a unique customer ID in this dataset, so repeat-purchase behavior cannot be assessed from the available records alone.

4. Conclusion

The analysis shows that Electronics and Clothing are the leading categories by sales revenue, while monthly sales vary considerably across the observed period. These findings provide a useful starting point for investigating category performance, understanding purchasing behavior, and making more informed retail decisions.
